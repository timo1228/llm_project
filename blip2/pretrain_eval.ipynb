```python
import torch

from blip2.custom import load_model
from blip2.custom.processors import load_processor
from PIL import Image
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

vis_processor = load_processor("blip2_image_eval").build(image_size=224)
model = load_model(name="blip2_llama", model_type="pretrain_llama3.2_3B", is_eval=True, device=device, checkpoint=r"D:\School\IntroToLLM\Project\output\BLIP2\Pretrain_stage2\20251202125\checkpoint_6.pth")

```

```python
image = Image.open(r"D:\cache\lavis\vg\images\3216.jpg").convert("RGB")
#image = Image.open("./data/cats.jpeg").convert("RGB")
image = vis_processor(image).unsqueeze(0).to(device)

res = model.generate({"image": image, "prompt": "A photo of"}, use_nucleus_sampling=True, max_length=128, temperature=0.5, repetition_penalty=2.0, num_captions=5, length_penalty=0.2)
print(res)
```
